# Hemodialyzer Separation Physics
This notebook demonstrates the mass transfer physical model, focusing on the separation of different solutes (Urea, Vitamin B12, Albumin) along the length of the dialysis membrane fibers.

```
import sys
import os
import numpy as np
import matplotlib.pyplot as plt

# Add src to path
sys.path.append(os.path.abspath('../src'))
from solver import solve_solute_profile
from parameters import SOLUTES, Q_b, Q_d, Q_uf, L
```

### 1. Solve Spatial Concentration Profiles
We solve the Boundary Value Problem corresponding to the 1D mass balances of blood and dialysate flow.

```
# Solve for Urea and VitB12
res_urea = solve_solute_profile("Urea", Q_b, Q_d, Q_uf)
res_b12 = solve_solute_profile("VitB12", Q_b, Q_d, Q_uf)
res_albumin = solve_solute_profile("Albumin", Q_b, Q_d, Q_uf)

print(f"Urea Clearance: {res_urea['Clearance_ml_min']:.2f} mL/min")
print(f"VitB12 Clearance: {res_b12['Clearance_ml_min']:.2f} mL/min")
print(f"Albumin Clearance: {res_albumin['Clearance_ml_min']:.2f} mL/min")
```

### 2. Plotting Concentration along the Fiber Length

```
plt.figure(figsize=(10, 6))

z_cm = res_urea['z'] * 100 # Convert to cm

# Urea
plt.plot(z_cm, res_urea['Cb_profile'], 'b-', lw=2, label='Urea (Blood)')
plt.plot(z_cm, res_urea['Cd_profile'], 'b--', lw=2, label='Urea (Dialysate)')

# VitB12
plt.plot(z_cm, res_b12['Cb_profile'], 'g-', lw=2, label='VitB12 (Blood)')
plt.plot(z_cm, res_b12['Cd_profile'], 'g--', lw=2, label='VitB12 (Dialysate)')

plt.title('Concentration Profiles along Hemodialyzer Fibers', fontsize=14)
plt.xlabel('Distance along fibers (cm)', fontsize=12)
plt.ylabel('Concentration (mg/mL)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
